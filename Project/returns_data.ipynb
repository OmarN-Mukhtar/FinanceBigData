```python
import pandas as pd
```

```python
df = pd.read_csv('data_sp500.csv')
```

```python
df.head()
```

```python
df['ret']
```

```python
df.sort_values('date', inplace=True)
```

```python
df.head()
```

```python
print(df['date'].dtype)
print(df.dtypes)
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

```python
print(df['date'].dtype)  
```

```python
ff = pd.read_csv('FF5.csv')
```

```python
ff.head()
```

```python
ff['Date'] = pd.to_datetime(ff['Date'], format='%Y%m%d', errors='coerce')
```

```python
ff.head()
```

```python
print(ff['Date'].dtype)
```

```python
ff.rename(columns={'Date': 'date'}, inplace=True)
```

```python
df = df.merge(ff, on='date', how='left', suffixes=('', '_ff'))

df.head()
```

```python
df['ew_ret'] = df.groupby('date')['ret'].transform('mean')
```

```python
df.head()
```

```python
df['ew_ret_excess'] = df['ew_ret'] - df['RF']
```

```python
df.head()
```

```python
df.rename(columns={'Mkt-RF': 'mkt-rf', 'SMB': 'smb', 'HML': 'hml', 'RMW': 'rmw', 'CMA': 'cma', 'RF': 'rf'}, inplace=True)
```

```python
df.to_csv('data_sp500_ff.csv', index=False)
```
